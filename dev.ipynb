```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, LabelEncoder, LabelBinarizer
from sklearn.linear_model import SGDClassifier
import matplotlib.pyplot as plt
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd
from src.config import PLAYER_DB_PATH, MATCHS_FILE
import sqlite3

# For each match, we will predict the winner
# Load the matches data
matchs_df = pd.read_csv(MATCHS_FILE, parse_dates=['start_datetime'])
# Load the players data
with sqlite3.connect(PLAYER_DB_PATH) as conn:
    players_df = pd.read_sql_query("SELECT * FROM players", conn)   
```

```python
def fetch_data(matchs_df, players_df):
    data = []

    for _, row in matchs_df.iterrows():
        rich_match = {}
        rich_match['match_id'] = row['match_id']
        rich_match['start_datetime'] = row['start_datetime']
        rich_match['winner'] = row['winner_team']
        
        # Collecter tous les PUUIDs des joueurs (5 bleus + 5 rouges)
        players_puuids = [row[f'blue_{i}_puuid'] for i in range(5)]
        players_puuids += [row[f'red_{i}_puuid'] for i in range(5)]
        
        # Récupérer les stats des joueurs
        players_stats = players_df[players_df['puuid'].isin(players_puuids)]
        
        # Vérifier qu'on a trouvé les stats de tous les joueurs
        if len(players_stats) != 10:
            # missing_puuids = set(players_puuids) - set(players_stats['puuid'].tolist())
            # print(f"Attention: Stats manquantes pour {len(missing_puuids)} joueurs dans le match {row['match_id']}")
            # print(f"PUUIDs manquants: {missing_puuids}")
            continue
        
        # Ajouter les stats des joueurs dans l'ordre des équipes (bleus puis rouges)
        for i, player in enumerate(players_stats.itertuples()):
            rich_match[f'player_{i}_puuid'] = player.puuid
            rich_match[f'player_{i}_name'] = player.name
            rich_match[f'player_{i}_level'] = player.level
            rich_match[f'player_{i}_match_count'] = player.match_count
            for stat in ['kills', 'deaths', 'assists', 'headshots', 'bodyshots', 'legshots']:
                rich_match[f'player_{i}_{stat}'] = getattr(player, stat, 0)
        
        # Convertir le dictionnaire en DataFrame
        data.append(rich_match) 
    return pd.DataFrame(data)
        
```

```python
class ValorantDataLoader:
    def __init__(self, match_dataframe, players_dataframe, data_fetch_fn, target_column="winner"):
        """
        data_fetch_fn: Callable that returns (X, y) as pandas DataFrames.
        target_column: Name of the target column in the data.
        """
        self.match_dataframe = match_dataframe
        self.players_dataframe = players_dataframe
        if not callable(data_fetch_fn):
            raise ValueError("data_fetch_fn must be a callable that returns (X, y) or a DataFrame with a target column")
        self.data_fetch_fn = data_fetch_fn
        self.target_column = target_column
        self.X, self.y = self._load_data()

    def _load_data(self):
        data = self.data_fetch_fn(self.match_dataframe, self.players_dataframe)
        if isinstance(data, tuple) and len(data) == 2:
            X, y = data
        elif isinstance(data, pd.DataFrame):
            y = data[self.target_column]
            X = data.drop(columns=[self.target_column])
        else:
            raise ValueError("data_fetch_fn must return either (X, y) or a DataFrame with a target column")
        # Label encode the target variable if it's categorical
        if y.dtype == 'object' or y.dtype.name == 'category':
            le = LabelEncoder()
            y = le.fit_transform(y)
        return X, y

    def get_data(self):
        return self.X, self.y

    def get_train_test_split(self, test_size=0.1, random_state=42):
        return train_test_split(self.X, self.y, test_size=test_size, random_state=random_state)
```

```python
loader = ValorantDataLoader(match_dataframe=matchs_df, players_dataframe=players_df, data_fetch_fn=fetch_data)
X, y = loader.get_data()
X_train, X_test, y_train, y_test = loader.get_train_test_split()
```

# Custom Preprocessing Script

```python
class RemoveUnwantedColumns:
    def __init__(self, columns_to_remove):
        self.columns_to_remove = columns_to_remove

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        return X.drop(columns=self.columns_to_remove)
```

```python
columns_to_remove = ['match_id', 'start_datetime']
columns_to_remove+= [f'player_{idx}_name' for idx in range(10)]  # Assuming 10 players (5 blue + 5 red)
columns_to_remove += [f'player_{idx}_puuid' for idx in range(10)]  # Assuming 10 players (5 blue + 5 red)
```

```python
pipeline = Pipeline([
    ('remove_unwanted_columns', RemoveUnwantedColumns(columns_to_remove=columns_to_remove)),
    ('scaler', StandardScaler()),
    ('classifier', SGDClassifier(loss='log_loss', penalty='l2', max_iter=10_000, tol=1e-3))
])

pipeline.fit(X_train, y_train) 
```

# Test the pipeline

```python
# Evaluate the model
score = pipeline.score(X_test, y_test)
print(f'Test score: {score:.4f}')
```

```python
#Confusion matrix
y_pred = pipeline.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print('Confusion Matrix:')

plt.figure(figsize=(8, 6))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.colorbar()
plt.xticks(np.arange(len(pipeline.classes_)), pipeline.classes_, rotation=45)
plt.yticks(np.arange(len(pipeline.classes_)), pipeline.classes_)
plt.xlabel('Predicted label')
plt.ylabel('True label')
plt.tight_layout()
plt.show()
```
